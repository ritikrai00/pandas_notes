```python
df1 = pd.DataFrame({
"city": ["new york","chicago","orlando"],
"temperature": [21,14,35]
})
df2 = pd.DataFrame({
"city": ["new york","chicago","san francisco"],
"humidity": [65,68,75]
})
print(df1)
print(df2)
```

```python
df3=pd.merge(df1,df2,on='city') #inner join
df3
```

```python
df3=pd.merge(df1,df2,on='city',how='outer') #union
df3
```

```python
df3=pd.merge(df1,df2,on='city',how='left') #left join
df3
```

```python
df3=pd.merge(df1,df2,on='city',how='right') #right join
df3
```

```python
df3=pd.merge(df1,df1,on='city') #self join
df3
```

```python
df3=pd.merge(df1,df2,how='cross') #cross join
df3
```

```python
df3=pd.merge(df1,df2,on='city',how='outer',indicator=True) #union
df3
```
